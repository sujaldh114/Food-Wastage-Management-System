#  Univariate Analysis

Univariate analysis focuses on examining one variable at a time. It helps in understanding the distribution and frequency of different categories in the datasets.

##  Chart 1: Provider Type Distribution

This chart represents the distribution of food listings contributed by different types of food providers. It helps identify which provider category contributes the highest number of food listings to the Food Wastage Management System.

```python
provider_type = food_listings["Provider_Type"].value_counts()

plt.figure(figsize=(5,5))
plt.bar(provider_type.index, provider_type.values)

plt.title("Provider Type Distribution")
plt.xlabel("Provider Type")
plt.ylabel("Number of Listings")

plt.xticks(rotation=20)

plt.show()
```

##  Observation

- The **Supermarket** category has the highest number of food listings, making it the largest contributor to the food donation system.
- **Restaurants** are the second-largest contributors, followed closely by **Grocery Stores**, indicating their significant role in reducing food wastage.
- **Catering Services** have the lowest number of food listings among all provider types.
- The distribution of food listings across provider types is relatively balanced, showing that food donations are received from multiple types of providers rather than relying on a single source.
- This balanced contribution helps improve the availability and diversity of food donations within the Food Wastage Management System.

# Chart 2: Food Type Distribution

This chart illustrates the distribution of different food types available in the food donation system. It helps identify which category of food is donated most frequently.

```python
food_type = food_listings["Food_Type"].value_counts()
colors = ['Green','Orange','Red']

plt.figure(figsize=(5,5))
plt.bar(food_type.index, food_type.values, color= colors)

plt.title("Food Type Distribution")
plt.xlabel("Food Type")
plt.ylabel("Number of Listings")

plt.xticks(rotation=15)

plt.show()
```

##  Observation

- **Vegetarian** food has the highest number of listings, making it the most commonly donated food category.
- **Vegan** food closely follows Vegetarian food, indicating a substantial availability of plant-based food donations.
- **Non-Vegetarian** food has the lowest number of listings among the three categories, although the difference is relatively small.
- The distribution of all three food types is fairly balanced, ensuring that receivers have access to a diverse range of food options.
- A balanced distribution of food types improves the inclusiveness and effectiveness of the food donation system.

##  Chart 3: Meal Type Distribution

This chart illustrates the distribution of different meal types available in the food donation system. It helps identify which meal category is donated most frequently.

```python
meal_type = food_listings["Meal_Type"].value_counts()

plt.figure(figsize=(5,5))
plt.pie(
    meal_type.values,
    labels=meal_type.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Meal Type Distribution")

plt.show()
```

##  Observation

- **Breakfast** has the highest share of food listings, accounting for approximately **25.4%** of the total donations.
- **Snacks** closely follow with **25.3%**, while **Lunch** and **Dinner** contribute **24.8%** and **24.5%**, respectively.
- The distribution of meal types is nearly uniform, indicating that food donations are well-balanced across all meal categories.
- Such a balanced distribution ensures that receivers have access to food suitable for different meal times throughout the day.
- The absence of a dominant meal type reflects diversity in the donated food inventory, improving the overall effectiveness of the food donation system.

## Chart 4: Distribution of Food Quantity

```python
plt.figure(figsize=(8,4))

plt.hist(
    food_listings["Quantity"],
    bins=15,
    color="Blue",
    edgecolor="black"
)

plt.title("Distribution of Food Quantity", fontsize=16)
plt.xlabel("Quantity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

### Insights

- The quantity of donated food ranges from approximately **1 to 50 units**.
- The histogram shows that food quantities are fairly evenly distributed across the dataset.
- No strong skewness is observed, indicating that both small and large quantity donations occur with similar frequency.
- There are no obvious outliers or unusual peaks, suggesting a well-balanced distribution of food quantities.

## Chart 5: Expiry Date Distribution

```python
expiry_counts = food_listings["Expiry_Date"].value_counts().sort_index()

plt.figure(figsize=(12,5))

plt.plot(
    expiry_counts.index,
    expiry_counts.values,
    marker="o",
    linewidth=2,
    color="Green"
)

plt.title("Food Listings by Expiry Date", fontsize=16)
plt.xlabel("Expiry Date")
plt.ylabel("Number of Listings")

plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
```

### Insights

- Food listings are distributed across expiry dates from **16 March 2025 to 30 March 2025**.
- The highest number of food listings occurs on **22 March 2025** and **24 March 2025**, with approximately **76 listings** each.
- The lowest number of food listings is observed on **21 March 2025**, with around **54 listings**.
- Overall, the number of food listings remains relatively consistent across the expiry dates, indicating a balanced distribution without any significant concentration on a single day.

# Bivariate Analysis

Bivariate analysis examines the relationship between two variables. It helps identify patterns, comparisons, and associations between different features in the dataset, providing deeper insights into how one variable influences another.

## Chart 6: Provider Type vs Total Quantity Donated

```python
provider_quantity = (
    food_listings
    .groupby("Provider_Type")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8,4))

plt.bar(
    provider_quantity.index,
    provider_quantity.values,
    color="Orange",
    edgecolor="black"
)

plt.title("Total Quantity Donated by Provider Type", fontsize=16)
plt.xlabel("Provider Type")
plt.ylabel("Total Quantity")

plt.xticks(rotation=15)

plt.tight_layout()
plt.show()
```

## Chart 7: Food Type vs Total Quantity Donated

```python
food_quantity = (
    food_listings
    .groupby("Food_Type")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(7,4))

colors = ['Orange','Red','Green']

plt.bar(
    food_quantity.index,
    food_quantity.values,
    color= colors,
    edgecolor="black"
)

plt.title("Total Quantity Donated by Food Type", fontsize=16)
plt.xlabel("Food Type")
plt.ylabel("Total Quantity")

plt.tight_layout()
plt.show()
```

### Insights

- Vegan food has the highest total donated quantity, contributing approximately **8,800 units**.
- Non-Vegetarian food follows closely with around **8,700 units**.
- Vegetarian food has the lowest donated quantity, with approximately **8,400 units**.
- The total donated quantity is fairly balanced across all food types, indicating that donations are well distributed among Vegan, Vegetarian, and Non-Vegetarian food categories.

## Chart 8: Meal Type vs Total Quantity Donated

```python
meal_quantity = (
    food_listings
    .groupby("Meal_Type")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(7,4))

plt.bar(
    meal_quantity.index,
    meal_quantity.values,
    color="Purple",
    edgecolor="black"
)

plt.title("Total Quantity Donated by Meal Type", fontsize=16)
plt.xlabel("Meal Type")
plt.ylabel("Total Quantity")

plt.tight_layout()
plt.show()
```

### Insights

- Snacks have the highest total donated quantity, contributing approximately **6,700 units**.
- Breakfast follows very closely with nearly **6,650 units**, showing almost equal contribution.
- Lunch contributes around **6,400 units**, making it the third highest meal category.
- Dinner has the lowest total donated quantity, with approximately **6,000 units**.
- The donation quantities across all meal types are relatively balanced, indicating that food donations are well distributed throughout different meal categories.

## Chart 9: Number of Food Listings by Provider Type

```python
provider_counts = food_listings["Provider_Type"].value_counts()

plt.figure(figsize=(8,5))

plt.bar(
    provider_counts.index,
    provider_counts.values,
    color="#2E8B57",
    edgecolor="black"
)

plt.title("Number of Food Listings by Provider Type", fontsize=16)
plt.xlabel("Provider Type")
plt.ylabel("Number of Listings")

plt.tight_layout()
plt.show()
```

### Insights

- Supermarkets have the highest number of food listings, with approximately **270 listings**.
- Restaurants rank second, contributing around **260 listings**.
- Grocery Stores have approximately **245 listings**, while Catering Services have the fewest at around **230 listings**.
- The difference between provider types is relatively small, indicating that all provider categories actively contribute food listings to the system.

## Chart 10: Average Quantity Donated by Food Type

```python
avg_quantity = (
    food_listings
    .groupby("Food_Type")["Quantity"]
    .mean()
    .sort_values(ascending=False)
)

colors = ['Orange','Red','Green']

plt.figure(figsize=(7,4))

plt.bar(
    avg_quantity.index,
    avg_quantity.values,
    color=colors,
    edgecolor="black"
)

plt.title("Average Quantity Donated by Food Type", fontsize=16)
plt.xlabel("Food Type")
plt.ylabel("Average Quantity")

plt.tight_layout()
plt.show()
```

### Insights

- Vegan and Non-Vegetarian food have the highest average donated quantity, each averaging approximately **26 units** per listing.
- Vegetarian food has a slightly lower average donation quantity of around **25 units**.
- The average donation quantity is very similar across all food types, indicating a balanced distribution of donation sizes.
- The small variation suggests that the quantity donated is largely independent of the food category.

## Chart 11: Number of Food Listings by Location (Top 10)

```python
top_locations = (
    food_listings["Location"]
    .value_counts()
    .head(10)
)

plt.figure(figsize=(10,5))

plt.bar(
    top_locations.index,
    top_locations.values,
    color="Brown",
    edgecolor="black"
)

plt.title("Top 10 Locations by Food Listings", fontsize=16)
plt.xlabel("Location")
plt.ylabel("Number of Listings")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

### Insights

- New Carol and South Kathryn have the highest number of food listings, with **6 listings** each.
- Jimmyberg, East Angela, and Perezport follow closely with **5 listings** each.
- The remaining top locations each contribute **4 listings**, indicating a fairly even geographical distribution.
- No single location dominates the dataset, suggesting that food donations are spread across multiple areas rather than being concentrated in one place.

## Chart 12: Number of Claims by Claim Status

```python
claim_status = claims["Status"].value_counts()

plt.figure(figsize=(7,4))

plt.bar(
    claim_status.index,
    claim_status.values,
    color="Purple",
    edgecolor="black"
)

plt.title("Number of Claims by Status", fontsize=16)
plt.xlabel("Claim Status")
plt.ylabel("Number of Claims")

plt.tight_layout()
plt.show()
```

### Insights

- Completed claims have the highest count, with approximately **340 claims**, indicating that most food requests are successfully fulfilled.
- Cancelled claims are almost equal to completed claims, suggesting that a significant number of food requests do not reach completion.
- Pending claims are the lowest, with around **325 claims**, indicating that relatively few requests remain unresolved.
- Overall, the distribution of claim statuses is fairly balanced, showing consistent activity across all stages of the claim process.
